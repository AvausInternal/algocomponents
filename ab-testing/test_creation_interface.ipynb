## Matric Creation
### Rules for defining the Matric
1. Matric definition needs to be specified in Sql
2. The martic should be names similar how you defined it in the test config file
3. Write a query that claculated the matric for each user included in the test, the matric will be calculated for the duration specified in the test config file
4. Matric should be on the customer level

```python
query = '''SELECT
SELECT
  customer_id,
  AVG(revenue) AS metric
FROM algo-factory-dev.online_retail.online_retail_clv_clean
WHERE order_date BETWEEN '{START_DATE}' AND '{END_DATE}'
GROUP BY customer_id
'''

metric = MetricCreator(
    metric_name='Average revenue',
    description='Average order revenue for each customer',
    sql=query
)
```
